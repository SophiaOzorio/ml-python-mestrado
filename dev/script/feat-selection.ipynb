```python
# Seleção de diretórios
import os

# Carregamento de dados
from joblib import dump, load

# Processamento de dados
import numpy as np
import pandas as pd

# Geração de gráficos
import matplotlib.pyplot as plt
import seaborn as sns

from itertools import combinations
```

```python
caminho_arquivo = "/home/sophia.ozorio/storage/python-ml/projeto_ML_python/dev/script/data/variantes-final.csv"

df_projeto = pd.read_csv(caminho_arquivo)
display(df_projeto.head())
```

```python
num_columns = df_projeto.select_dtypes(include="number").columns
num_columns
```

```python
eixos_graficos = list(combinations(num_columns, 2))

print(f"Total de pares: {len(eixos_graficos)}")
print(eixos_graficos)
```

```python
for c1, c2 in eixos_graficos:

    print("-" * 60)
    print(f"{c1} × {c2}")

    p = sns.jointplot(
        x=c1,
        y=c2,
        hue="ataxia",
        height=6,
        ratio=4,
        marginal_ticks=True,
        data=df_projeto,
    )

    p.fig.suptitle(
        f"{c1} × {c2} por classe de ataxia",
        y=1.02
    )

    p.set_axis_labels(
        xlabel=c1,
        ylabel=c2
    )

    plt.show()
```
