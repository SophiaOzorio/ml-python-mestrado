```python
import os
import pandas as pd
from ydata_profiling import ProfileReport
```

```python

caminho_dir = "/home/sophia.ozorio/storage/python-ml/projeto_ML_python/dev/script/results/split_dataset/"
caminho_relatorios = "/home/sophia.ozorio/storage/python-ml/projeto_ML_python/dev/script/results/reports/"

arquivos = [
    "X_train_ready.csv",
    "X_test_ready.csv",
    "X_valid_ready.csv",
    "y_ataxia_train_ready.csv",
    "y_ataxia_test_ready.csv",
    "y_ataxia_valid_ready.csv",
    "y_idade_train_ready.csv",
    "y_idade_test_ready.csv",
    "y_idade_valid_ready.csv"
]

for nome_arquivo in arquivos:
    caminho_csv = os.path.join(caminho_dir, nome_arquivo)
    df = pd.read_csv(caminho_csv)

    nome_base = nome_arquivo.replace(".csv", "")
    print(f"--- Gerando profiling: {nome_base} ({df.shape[0]} linhas, {df.shape[1]} colunas) ---")

    profile = ProfileReport(df, title=f"Profiling Report - {nome_base}", minimal=True)
    profile.to_file(os.path.join(caminho_relatorios, f"{nome_base}_report.html"))

print("Todos os relatórios gerados com sucesso!")
```
