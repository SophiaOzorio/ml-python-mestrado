```python
import pandas as pd
from joblib import dump

df = pd.read_csv("/home/sophia.ozorio/storage/python-ml/projeto_ML_python/dev/script/data/variantes-final.csv")

categoricas = ["gene", "tipo_variante", "ancestralidade", "especie",
               "genoma_referencia", "classificacao_clinica", "ataxia"]
booleanas = ["em_dominio_funcional", "historico_familiar"]

for c in categoricas:
    df[c] = df[c].astype("category")
for c in booleanas:
    df[c] = df[c].map({"True": True, "False": False, True: True, False: False}).astype("boolean")

dump(df, "/home/sophia.ozorio/storage/python-ml/projeto_ML_python/dev/script/data/variantes-final.plk")
```
